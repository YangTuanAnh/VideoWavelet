```python
import os
import numpy as np
import pandas as pd
import glob
import subprocess
from pathlib import Path
from PIL import Image

import torch
import open_clip
from faster_whisper import WhisperModel, BatchedInferencePipeline
```

```python
VIDEO_FOLDER_ID = "Videos_L22_a"
VIDEO_DIR = f"/kaggle/input/datasets/yangtunanh/aic-2026-l25-a/{VIDEO_FOLDER_ID}/video"
FRAMES_DIR = "frames"
SCENES_DIR = "scenes"
SUBTITLES_DIR = "subtitles"
EMB_DIR = "embeddings"

BATCH_SIZE = 64
CLIP_MODEL = "ViT-B-32-SigLIP2-256"
PRETRAINED_WEIGHTS = "webli"
video_ext = ["*.mp4", "*.mkv", "*.avi", "*.mov", "*.webm", "*.flv"]

device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
files = []
for ext in video_ext:
    files.extend(glob.glob(os.path.join(VIDEO_DIR, ext)))

for f in sorted(files):
    print("Processing:", f)

    subprocess.run([
        "scenedetect",
        "-i",
        f,
        "-fs", "24",
        "-q",
        "list-scenes",
        "-q",
        "-o", SCENES_DIR,
        "save-images",
        "-o", FRAMES_DIR,
        "-H", "448", "-W", "448",
        "-n", "1",
        "-f", "$VIDEO_NAME/Scene-$SCENE_NUMBER-$FRAME_NUMBER",
    ])
```

```python
model, preprocess = open_clip.create_model_from_pretrained(CLIP_MODEL, PRETRAINED_WEIGHTS)
model.eval().to(device)
tokenizer = open_clip.get_tokenizer(CLIP_MODEL)

os.makedirs(EMB_DIR, exist_ok=True)

with torch.no_grad(), torch.autocast(device):
    for vid in os.listdir(FRAMES_DIR):
        vid_path = os.path.join(FRAMES_DIR, vid)
        print("Embedding:", vid_path)
        emb_path = os.path.join(EMB_DIR, f"{vid}.npy")

        image_files = sorted(os.listdir(vid_path))
        embeddings = []
        
        for i in range(0, len(image_files), BATCH_SIZE):
            batch_imgs = []
        
            for img_name in image_files[i:i+BATCH_SIZE]:
                img = Image.open(os.path.join(vid_path, img_name)).convert("RGB")
                batch_imgs.append(preprocess(img))
        
            batch = torch.stack(batch_imgs).cuda()
        
            image_features = model.encode_image(batch)
            image_features = image_features / image_features.norm(dim=-1, keepdim=True)
        
            embeddings.append(image_features.cpu().float().numpy())
        
        embeddings = np.concatenate(embeddings, axis=0)
        np.save(emb_path, embeddings)

print("Done with embeddings!")
```

```python
files = []
for ext in video_ext:
    files.extend(glob.glob(os.path.join(VIDEO_DIR, ext)))

os.makedirs(SUBTITLES_DIR, exist_ok=True)

model_size = "turbo"
model = WhisperModel(model_size, device=device)
batched_model = BatchedInferencePipeline(model=model)

for video_path in sorted(files):
    stem = Path(video_path).stem
    audio_path = "tmp.wav"

    print("Transcribing:", video_path)

    subprocess.run([
        "ffmpeg",
        "-y",
        "-i", video_path,
        "-hide_banner", "-loglevel", "error",
        "-vn",
        "-ac", "1",
        "-ar", "16000",
        "-acodec", "pcm_s16le",
        audio_path
    ], check=True)

    segments, info = batched_model.transcribe(audio_path, beam_size=5, language="vi", batch_size=16)

    data = []
    for segment in segments:
        data.append({
            "start_time": segment.start,
            "end_time": segment.end,
            "text": segment.text
        })
    df = pd.DataFrame(data)

    subtitle_path = os.path.join(SUBTITLES_DIR, f"{stem}-Subtitles.csv")
    df.to_csv(subtitle_path, encoding="utf-8")
```
